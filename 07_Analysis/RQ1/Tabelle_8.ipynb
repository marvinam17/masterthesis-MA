# Analysis of Query differences

```python
from beir.datasets.data_loader import GenericDataLoader
import numpy as np
import statistics
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
corpus, queries, qrels = GenericDataLoader(data_folder="../../03_data/raw_data/beir_dbpedia/dbpedia-entity").load(split="test")
```

```python
def differ_query_types(query):
    if query.startswith("INEX_LD"):
        return "INEX-LD"
    elif query.startswith("QALD2"):
        return "QALD-2"
    elif query.startswith("SemSearch_ES"):
        return "SemSearch ES"
    else:
        return "ListSearch"
    
query_types = {}
for key in queries.keys():
    query_type = differ_query_types(key)
    if query_type in query_types.keys() and isinstance(query_types[query_type], list):
        query_types[query_type].append(key)
    else:
        query_types[query_type] = [key]
```

```python
print("Inex:", queries["INEX_LD-2010004"])
print("QALD:", queries["QALD2_tr-8"])
print("SemSearch:", queries["SemSearch_ES-33"])
print("ListSearch:", queries["TREC_Entity-7"])
```

```python
# Tabelle 8
for category, query_set in query_types.items():
    stopword_rate = []
    for query in query_set:
        word_tokens = word_tokenize(queries[query])
        stopwords_q =  [w for w in word_tokens if w in stopwords.words('english')]
        stopword_rate.append(len(stopwords_q) / len(word_tokens))
    print(category, np.mean(stopword_rate))
```

```python
# Tabelle 8
query_types = {}
for key in queries.keys():
    query_type = differ_query_types(key)
    if query_type in query_types.keys() and isinstance(query_types[query_type], list):
        query_types[query_type].append(len(word_tokenize(queries[key])))
    else:
        query_types[query_type] = [len(word_tokenize(queries[key]))]

for key in query_types.keys():
    print(key, ":", statistics.mean(query_types[key]))
```
